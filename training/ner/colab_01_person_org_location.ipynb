# Bitirme Maskeleme: ilk NER eğitimi
Bu not defteri gerçek metin bağlamı içeren WikiNER verisiyle kişi, kurum ve yer adlarını tanıyan ilk BERTurk modelini Colab GPU’da eğitir. Açık adresler ve sayısal veriler ayrı kural katmanlarında ele alınacaktır.

## Başlamadan önce
1. Colab’da **Dosya → Not defterini yükle** ile bu `.ipynb` dosyasını aç.
2. **Çalışma zamanı → Çalışma zamanı türünü değiştir → T4 GPU** seç (listede varsa).
3. İlk hücre çalışınca bilgisayarından `person-org-location-colab.zip` dosyasını seç.


```python
from google.colab import files

uploaded = files.upload()  # person-org-location-colab.zip dosyasını seç
archive_name = "person-org-location-colab.zip"
if archive_name not in uploaded:
    raise FileNotFoundError(f"{archive_name} yüklenmedi")

import zipfile

with zipfile.ZipFile(archive_name) as zf:
    zf.extractall("/content")
print("Proje /content/bitirme_maskeleme içine açıldı")
```
